## Part 7: Two ways to divide by ten

In ordinary arithmetic `a / 10` and `a * (1/10)` are the same thing.

1. Compute Greendale's math/science average with `/`.
2. Compute it again with `math.prod([total, 1/10])` — `math.prod()` multiplies everything in a list.
3. Print both, then compare with `==` and with the `1e-13` test.

The data did not change. Only the route did.

````python
# your code here

average_A = sum_greendale_math / 10
average_B = math.prod([sum_greendale_math, 1/10])

print(average_A, average_B)
print(average_A == average_B)
print(abs(average_A - average_B) < 1e-13)
````

**So where is the error coming from?**

*Multiplication versus division, 1/10 also automatically becomes .1*
