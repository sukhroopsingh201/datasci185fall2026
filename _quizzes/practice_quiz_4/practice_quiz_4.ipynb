$$ \textrm{speed in meters per second} = \frac{1609.34}{3600} \times \textrm{speed in miles per hour}$$

3. Add a new column to the DataFrame named "Kinetic Energy (J)" which contains the kinetic energy of the car. The formula for kinetic energy is given below, where $m$ is the mass in kilograms and $v$ is the velocity in meters per second.

**Question (D)** A method that changes an object

<font size = "3">

The Emory Run Club meets on the WoodPEC outdoor track. The cell below defines a `Runner` class. Each runner has a `name`, a `stamina` between 0 and 100, and a count `num_runs` of the runs they have logged. The class already has getters, a setter `set_stamina` that uses `assert` to keep stamina between 0 and 100, and a method `set_pace_buddy(self, other)` that makes two runners each other's pace buddy. 

**Your task**

Add a method `log_run(self, score)` to the class. `score` is how hard the run was, from 0 to 100. The method should:

- use `assert` to make sure `log_run` only accepts scores between 0 and 100,
- add 1 to `num_runs`, and
- increase the runner's stamina by a tenth of `score`, but never above 100. Use the appropriate setter rather than calling the attribute directly to change the stamina.

**For example:**

```python
ana = Runner("Ana", 50)
ana.log_run(80)
print(ana.get_stamina(), ana.get_num_runs())

ben = Runner("Ben", 95)
ben.log_run(80)
print(ben.get_stamina())
```
should print 58.0 1 and then 100 (not 103.0).

After the test cell, run the cell below it. It should fail with your `assert` message.

````python
# Add the method log_run to the Runner class below

class Runner:
    """A member of the Emory Run Club."""

    def __init__(self, name, stamina):
        self.name = name
        self.stamina = stamina
        self.num_runs = 0
        self.pace_buddy = None

    def __str__(self):
        return f"{self.name} (stamina {round(self.stamina, 1)})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_stamina(self):
        return self.stamina

    def get_num_runs(self):
        return self.num_runs

    def get_pace_buddy(self):
        return self.pace_buddy

    # ---------- setters ----------

    def set_stamina(self, value):
        assert 0 <= value <= 100, "stamina must be between 0 and 100"
        self.stamina = value

    def set_pace_buddy(self, other):
        """Make self and other each other's pace buddy."""
        self.pace_buddy = other
        other.pace_buddy = self

    # ---------- scoring ----------

    def run_value(self, score):
        return score

    # ***************************************************
    # ---------- your code here: write log_run ----------
    # ***************************************************

    def log_run(self, score):
        assert score >= 0 and score <= 100, "Score cannot be below 0 or above 100"
        self.num_runs = self.num_runs + 1

        stamina = self.stamina + score / 10 
        if self.stamina + score / 10 > 100:
            stamina = 100
        self.set_stamina(stamina)
````

````python
# test cell, do not change 
ana = Runner("Ana", 50)
ana.log_run(80)
print(ana.get_stamina(), ana.get_num_runs())

ben = Runner("Ben", 95)
ben.log_run(80)
print(ben.get_stamina())
````

**Question (F)** Swoop picks pace buddies

<font size = "3">

The Emory Run Club meets on the WoodPEC outdoor track. Before each run, Swoop pairs up runners as pace buddies, as long as their staminas are close enough. The set-up cell below runs the script `quiz_data/q4_set_up_DO_NOT_CHANGE.py`, which defines the same `Runner` class as Question (A). The script also creates a dictionary `roster` of eight runners. **`roster` changes every time you run the set-up cell**, so make sure your code works for any roster. The test cell uses a fixed roster, so its output is the same every time.

**Your task:**

Write a class `RunClub` that holds the club's runners. It should have:


- a class variable `max_gap` equal to `10`. Swoop won't pair two runners whose stamina differs by more than `max_gap`.
- a constructor `__init__(self, roster)`, where `roster` is a dictionary like the one printed by the set-up cell. It should loop over the keys in the dictionary and store a `Runner` object for each name in a list attribute called `runners`.
- a method `swoop_set_buddys(self)` that pairs up the runners according to the following procedure: 
    - The method should loop over `runners` in order. 
    - For each runner who doesn't have a pace buddy yet, find the **first** other runner in `runners` who also doesn't have a pace buddy and whose stamina is within `max_gap` of theirs (a difference of exactly `max_gap` is allowed). 
        - Then use `set_pace_buddy` to make the two of them each other's pace buddy. If no one fits, that runner stays without a pace buddy. The method doesn't need to return anything.

*Hint 1:* You'll need a `for` loop inside a `for` loop. 

*Hint 2:* `abs(x - y)` gives the distance between two numbers.

**For example:**

```python
club = RunClub({"Ana": 50, "Ben": 95, "Cy": 56, "Dee": 47})
club.swoop_set_buddys()
for runner in club.runners:
    print(runner, "is paired with", runner.get_pace_buddy())
```
should print:

```text
Ana (stamina 50) is paired with Cy (stamina 56)
Ben (stamina 95) is paired with None
Cy (stamina 56) is paired with Ana (stamina 50)
Dee (stamina 47) is paired with None
```

````python
# set-up cell, do not change
# (run this cell before working on Question (F))

from quiz_data.q4_set_up_DO_NOT_CHANGE import *

print("roster:", roster)
````

````python
# Define the RunClub class here

class RunClub:
    """The Emory Run Club, which meets on the WoodPEC outdoor track."""
    max_gap = 10

    def __init__(self, roster):
        self.runners = []
        for runner in roster:
            self.runners.append(Runner(runner, roster[runner]))

    def swoop_set_buddys(self):
        for i in range(len(self.runners)):
            if self.runners[i].pace_buddy == None:
                for j in range(i + 1, len(self.runners)):
                    if self.runners[i].pace_buddy != None:
                        break
                    if self.runners[j].pace_buddy == None and abs(self.runners[i].stamina - self.runners[j].stamina) <= self.max_gap:
                        self.runners[i].pace_buddy = self.runners[j]
                        self.runners[j].pace_buddy = self.runners[i]
````

````python
# test cell, do not change
test_roster = {"Kiara": 71, "Xavier": 83, "Olga": 93, "Kai": 76, "Ines": 70, "Hugo": 45}

club = RunClub(test_roster)
club.swoop_set_buddys()

for runner in club.runners:
    print(runner, "is paired with", runner.get_pace_buddy())

# expected output:
# Kiara (stamina 71) is paired with Kai (stamina 76)
# Xavier (stamina 83) is paired with Olga (stamina 93)
# Olga (stamina 93) is paired with Xavier (stamina 83)
# Kai (stamina 76) is paired with Kiara (stamina 71)
# Ines (stamina 70) is paired with None
# Hugo (stamina 45) is paired with None
````
